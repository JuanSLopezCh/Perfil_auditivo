```python
import pandas as pd
import ast

df = pd.read_csv('../Data/Processed/Dataset_limpio_2.csv')

df.head()
```

```python
df["Frecuencia"] = df["Frecuencia"].apply(ast.literal_eval)
df["Respuesta"] = df["Respuesta"].apply(ast.literal_eval)
```

```python
Columnas1 = ["Evaluador", "TipoAudifono", "MarcaReferencia", "Marca","Referencia" ]
Columnas2 = ["MarcaReferencia", "Frecuencia", "Respuesta"]
```

```python
df_unicos = df[Columnas1].drop_duplicates()
df_listas = df[Columnas2]
```

```python
df_expandido = df_listas.apply(
    lambda row: pd.DataFrame({
        "MarcaReferencia": row["MarcaReferencia"],
        "Frecuencia": row["Frecuencia"],
        "Respuesta": row["Respuesta"]
    }),
    axis=1
).reset_index(drop=True)
```

```python
df_resultado = pd.concat(df_expandido.tolist(), ignore_index=True)
```

```python
df_resultado.to_csv('../Data/Processed/Dataset_Expandido.csv', index=False)
```

```python
df_unicos.to_csv('../Data/Processed/Dataset_Unicos.csv', index=False)
```
