```python
import pandas as pd

df = pd.read_csv('../Data/Processed/DataSet_limpio_2.csv')

df_clean = df['MarcaReferencia']
df_clean.head()
```

```python
def split_and_save(dataframe, chunk_size=100, output_prefix='Cluster_'):
    """
    Divide un DataFrame en chunks y los guarda como archivos CSV separados.
    
    Parámetros:
    - dataframe: DataFrame de pandas a dividir
    - chunk_size: número de registros por archivo (100 por defecto)
    - output_prefix: prefijo para los nombres de archivo
    """
    total_rows = len(dataframe)
    num_files = (total_rows // chunk_size) + (1 if total_rows % chunk_size else 0)
    
    for i in range(num_files):
        start_idx = i * chunk_size
        end_idx = start_idx + chunk_size
        chunk = dataframe.iloc[start_idx:end_idx]
        
        # Crear nombre de archivo (ej: MarcaReferencia_1.csv)
        filename = f"{output_prefix}{i+1}.csv"
        chunk.to_csv(f'../Data/Clusters/{filename}', index=False)
        print(f"Guardado {len(chunk)} registros en {filename}")
```

```python
split_and_save(df_clean, chunk_size=100)
```
